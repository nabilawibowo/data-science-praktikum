```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Dataset dengan missing value & duplikasi

data = {
    'Nama': ['Andi', 'Budi', 'Citra', 'Dina', 'Eko', 'Andi'],
    'Umur': [20, 21, np.nan, 22, 20, 20],
    'IPK': [3.5, 3.7, 3.2, np.nan, 3.6, 3.5]
}

df = pd.DataFrame(data)
df
```

```python
# Cek info dan missing value

print("Info Data:")
df.info()

print("\nMissing Value:")
print(df.isnull().sum())
```

```python
# Mengisi nilai kosong dengan rata-rata

df['Umur'].fillna(df['Umur'].mean(), inplace=True)
df['IPK'].fillna(df['IPK'].mean(), inplace=True)

df
```

```python
# Hapus data duplikat

df = df.drop_duplicates()
df
```

```python
# Statistik deskriptif

df.describe()
```

```python
df.plot(x='Nama', y='IPK', kind='bar', title='IPK Mahasiswa')
plt.show()
```

```python
df['Umur'].plot(kind='hist', bins=5, title='Distribusi Umur')
plt.show()
```

```python
df.plot(x='Umur', y='IPK', kind='scatter', title='Hubungan Umur dan IPK')
plt.show()
```

```python
# Korelasi

correlation = df.corr(numeric_only=True)
print("Korelasi:\n", correlation)
```
