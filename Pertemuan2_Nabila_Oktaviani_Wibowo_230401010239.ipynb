```python
# Import library
import pandas as pd
import numpy as np
```

```python
# Membuat dataset sederhana

data = {
    'Nama': ['Andi', 'Budi', 'Citra', 'Dina', 'Eko'],
    'Umur': [20, 21, 19, 22, 20],
    'IPK': [3.5, 3.7, 3.2, 3.8, 3.6]
}

df = pd.DataFrame(data)
df
```

```python
# Melihat informasi dataset

print("Info Data:")
df.info()

print("\nStatistik Data:")
df.describe()
```

```python
# Mengakses kolom tertentu

df['Nama']

# Mengakses baris tertentu
df.iloc[0]

# Mengakses data berdasarkan kondisi
df[df['IPK'] > 3.5]
```

```python
# Menambahkan kolom baru (Status Kelulusan)

df['Status'] = df['IPK'] >= 3.5
df
```

```python
# Mengurutkan data berdasarkan IPK

df_sorted = df.sort_values(by='IPK', ascending=False)
df_sorted
```

```python
import matplotlib.pyplot as plt

# Plot IPK
df.plot(x='Nama', y='IPK', kind='bar', title='IPK Mahasiswa')
plt.show()
```
