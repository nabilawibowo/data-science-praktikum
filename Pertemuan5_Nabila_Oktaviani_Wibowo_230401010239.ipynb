```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset('iris')

sns.scatterplot(
    data=df,
    x='sepal_length',
    y='petal_length',
    hue='species'
)

plt.show()
```

```python
sns.histplot(
    data=df,
    x='sepal_length',
    kde=True
)

plt.show()
```

```python
sns.boxplot(
    data=df,
    x='species',
    y='petal_length'
)

plt.show()
```

```python
sns.pairplot(df, hue='species')
plt.show()
```

```python
fig, axes = plt.subplots(2,2, figsize=(12,8))

sns.histplot(df['sepal_length'], ax=axes[0,0])
sns.boxplot(data=df, x='species', y='petal_length', ax=axes[0,1])
sns.scatterplot(data=df, x='sepal_length', y='petal_length', ax=axes[1,0])
sns.violinplot(data=df, x='species', y='sepal_width', ax=axes[1,1])

plt.tight_layout()
plt.show()
```
